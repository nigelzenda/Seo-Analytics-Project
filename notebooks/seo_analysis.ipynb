```python
pages_df["CTR"] = pages_df["CTR"].str.replace("%", "").astype(float)
```

```python
pages_df["CTR"] = pages_df["CTR"] / 100
```

```python
pages_df.info()
```

```python
pages_df.head()
```

```python
pages_df.isnull().sum()
```

```python
pages_df[pages_df["Clicks"] == 0]
```

```python
zero_click_pages = (pages_df["Clicks"] == 0).sum()
zero_click_pages
```

```python
total_pages = len(pages_df)
total_pages
```

```python
zero_click_percentage = (zero_click_pages / total_pages) * 100
zero_click_percentage
```

```python
pages_df[pages_df["Clicks"] > 0]
```

```python
pages_df["Position"].corr(pages_df["Clicks"])
```

```python
import matplotlib.pyplot as plt

plt.scatter(pages_df["Position"], pages_df["Clicks"])

plt.xlabel("Average Search Position")
plt.ylabel("Clicks")
plt.title("Search Position vs Clicks")

plt.show()
```

```python
pages_df.sort_values("Impressions", ascending=False)
```

```python
pages_df.sort_values("Impressions", ascending=False)
```

```python
opportunities = pages_df[
    (pages_df["Impressions"] > 0) &
    (pages_df["Clicks"] == 0)
]

opportunities
```

```python
opportunity_impressions = opportunities["Impressions"].sum()

opportunity_impressions
```

```python
total_impressions = pages_df["Impressions"].sum()

total_impressions
```

```python
opportunity_percentage = (opportunity_impressions / total_impressions) * 100

opportunity_percentage
```

```python
round(opportunity_percentage, 2)
```
